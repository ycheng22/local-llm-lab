## 2. Update Configuration
Modify `configs/sft_v1.yaml` to point to the new 5K datasets.

```python
import yaml

with open('../configs/sft_v1.yaml', 'r') as f:
    config = yaml.safe_load(f)
    
config['training']['dataset'] = 'data/train_5k.jsonl'
config['training']['eval_dataset'] = 'data/dev_500.jsonl'

with open('../configs/sft_v1_5k.yaml', 'w') as f:
    yaml.dump(config, f)
```

## 3. Run SFT Training
This will take several hours on an 8GB GPU.

```python
import sys
import os
sys.path.append(os.path.abspath("../src"))
sys.argv = ["sft.py", "--config", "../configs/sft_v1_5k.yaml"]
from llm_lab.training.sft import main
main()
```
