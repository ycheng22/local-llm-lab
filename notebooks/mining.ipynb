# Phase 4: Hard Example Mining
This notebook runs the data mining script using `configs/mining.yaml` to identify hard tasks.

We run `miner.py` natively through the Python kernel to avoid Windows CLI buffering and see the tqdm progress bar stream live.

```python
import os
import sys

# Ensure prints are flushed to avoid buffering delays on Windows
import builtins
if not hasattr(builtins, '__original_print__'):
    builtins.__original_print__ = builtins.print

def flush_print(*args, **kwargs):
    kwargs['flush'] = True
    builtins.__original_print__(*args, **kwargs)
builtins.print = flush_print

os.environ["CUDA_VISIBLE_DEVICES"] = "0"
os.environ["TOKENIZERS_PARALLELISM"] = "false"

sys.path.append(os.path.abspath("../src"))
sys.argv = ["miner.py", "--config", "../configs/mining.yaml"]

from llm_lab.data.miner import main
main()
```
